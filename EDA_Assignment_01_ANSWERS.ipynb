## **Nulls** [8 marks]

[1+1+3+2+1]

Find out how many nulls in each column

```python
sales.isnull().sum()
```

Find all those rows where brand is null

```python
sales[sales["brand"].isnull()]
```

Fill all the nulls in brand by extracting the brand name from product_name

(hint: look into str.rsplit)

```python
sales["brand"] = sales["brand"].fillna(
    sales["product_name"].str.rsplit(" ", n=1).str[-1]
)
```

Fill all the nulls in payment_method with the most frequently occuring payment method of the WHOLE column

```python
most_freq = sales["payment_method"].mode()[0]
sales["payment_method"] = sales["payment_method"].fillna(most_freq)
```

Check total nulls specifically in the payment_method column to see if all were successfully removed

```python
sales["payment_method"].isnull().sum()
```

## **Messy Data** [5 marks]

[1 + 3 + 1]

Find all the unique subcategories

```python
sales['&subcategory'].unique()
```

Clean up the messy text in the subcategory column

(hint: look for a .str function for this)

```python
sales['&subcategory'] = sales['&subcategory'].str.strip().str.lower()
```

Rename the subcategory column to remove messy text from there as well

```python
sales.rename(columns={"&subcategory": "clean_subcategory"}, inplace=True)
```

### **Total Sales** [3 marks]

Create a total_sale column (remember to factor in discount). It should be rounded to 2 decimal places

```python
sales['total_sale'] = (sales['quantity'] * sales['price'] * (1 - sales['discount_percent'])).round(2)
```

### **Groupby and Transform** [6 marks]

[3 + 3]

For each null in the price column, replace it with the average price of the subcategory occuring at that row.

(hint: use transform to get a column of means that is mappable to our dataframe and use that with fillna)


```python
sales['price'] = sales['price'].fillna(
    sales.groupby('clean_subcategory')['price'].transform('mean')
)
```

Find out how many unique brands in each subcategory.

Store this result in a variable called brands_per_subcategory

```python
brands_per_subcategory = sales.groupby('clean_subcategory')['brand'].nunique()
brands_per_subcategory
```

Plot a count plot of brands.

In the above plot, try and rearrange the bars in **ascending** order by height (i.e. by count)

```python
brand_order = sales['brand'].value_counts().sort_values().index

plt.figure(figsize=(12,6))
sns.countplot(data=sales, x='brand', order=brand_order)
plt.xticks(rotation=90)
plt.show()
```
